```python
import pandas as pd
import matplotlib.pyplot as plt

# URL do dataset real de e-commerce
url_real = 'https://raw.githubusercontent.com/gustav0moraes/datasets/main/online_retail_sample.csv'

try:
    df = pd.read_csv(url_real)
    print("Dados reais carregados com sucesso!\n")
except Exception as e:
    print("Não foi possível carregar a URL externa. Usando dados locais equivalentes para demonstração.\n")
    # Fallback estruturado exatamente com as mesmas colunas do dataset real
    dados_locais = {
        'Description': ['Notebook', 'Celular', 'Teclado', 'Mouse', 'Monitor', 'Notebook', 'Teclado', 'Monitor'],
        'Quantity': [2, 5, 10, 15, 3, 1, 4, 2],
        'UnitPrice': [3500.0, 1500.0, 150.0, 80.0, 900.0, 3500.0, 150.0, 900.0],
        'Country': ['Brasil', 'Brasil', 'Portugal', 'Portugal', 'Espanha', 'Brasil', 'Espanha', 'Portugal']
    }
    df = pd.DataFrame(dados_locais)

# Cálculo do faturamento real (Quantidade x Preço Unitário)
df['Faturamento'] = df['Quantity'] * df['UnitPrice']

# Agrupando faturamento por país (Top 5)
faturamento_por_pais = df.groupby('Country')['Faturamento'].sum().sort_values(ascending=False).head(5)
print("Faturamento por País (Top 5):\n", faturamento_por_pais, "\n")

# Gráfico de barras simples
faturamento_por_pais.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Top 5 Países por Faturamento')
plt.xlabel('País')
plt.ylabel('Faturamento ($)')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# Faturamento total geral do e-commerce
faturamento_total = df['Faturamento'].sum()
print(f"Faturamento Total Geral: $ {faturamento_total:,.2f}")

# Produtos mais vendidos (por faturamento)
faturamento_por_produto = df.groupby('Description')['Faturamento'].sum().sort_values(ascending=False).head(5)
print("\nTop 5 Produtos por Faturamento:\n", faturamento_por_produto)
```
